```python
import pandas as pd
import numpy as np
import random
```

```python
job_roles = ["Data Analyst", "Software Engineer", "Data Scientist", "Web Developer", "ML Engineer"]

cities = ["Mumbai", "Bangalore", "Pune", "Delhi", "Hyderabad"]

skills = ["Python", "SQL", "Excel", "Machine Learning", "Power BI", "Java", "React"]

companies = ["TCS", "Infosys", "Google", "Amazon", "Wipro", "Accenture"]

data = []

for i in range(500):
    data.append([
        random.choice(job_roles),
        random.choice(companies),
        random.choice(cities),
        random.randint(300000, 2500000),
        random.choice(skills),
        random.choice(["Fresher", "Mid", "Senior"]),
        pd.to_datetime("2023-01-01") + pd.to_timedelta(random.randint(0, 600), unit="D")
    ])

df = pd.DataFrame(data, columns=[
    "Job Title", "Company", "Location", "Salary", "Skill", "Experience", "Date"
])

df.head()
```

```python
df.to_csv("data/jobs.csv", index=False)
```
